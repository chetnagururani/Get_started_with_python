# Control Flow Statement
- Control flow statements allow the code to make decisions, iterate over data, and handle unexpected situations, creating dynamic and responsive programs.

1. conditional: if-else
2. Iterative: for and while
3. Jump: break, pass, continue

## Conditional Statement
- Conditional statements in Python allow you to execute a block of code only if a specified condition is true. These are fundamental in programming because they enable decision-making in code
#### Types of Conditional Statements
1. <b>if Statement
- They are used to execute a code block only if a specified condition is true.

```python
a=4
if a>=4:
    print("yes")
```

```python
# Short-handed if
a=9
b=8
if a > b: print("a is greater than b")
```

2. <b> if-else Statement
- Provides an alternative block of code that runs if the condition in the if statement is false. 

```python
# Example:
python
Copy code
age = 16
if age >= 18:
    print("You are eligible to vote.")
else:
    print("You are not eligible to vote.")
```

```python
# short-handed if -else
a = 2
b = 330
print("A") if a > b else print("B")
```

3. <b> if-elif-else Statement
- Allows checking multiple conditions. Once a condition is met, the corresponding block of code is executed, and the rest are ignored.

Syntax:
if condition1:
    # code block if condition1 is true
elif condition2:
    # code block if condition2 is true
else:
    # code block if none of the conditions are true

```python
# Example:
score = 85
if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: D")
```

```python
#elif
a= int(input())
if a<=20:
    print('less than 20')
elif(a==20):
    print("equal to 20")
elif(a>20 and a<50):
    print("value is greater than 20 and less than 50")
else: 
    print("greater than 90")
    
```

4. <b> Nested if
- You can place an if statement inside another if statement to check multiple conditions.

```python
Example:
num = 10
if num > 0:
    print("Positive number")
    if num % 2 == 0:
        print("Even number")
```

# While loop

- With the while loop we can execute a set of statements as long as a condition is true.
- A while loop in programming repeatedly executes a block of code as long as a specified condition remains true. It's useful when the number of iterations is not known beforehand and is determined by a condition during execution.
- remember to increment i, or else the loop will continue forever.

!

```python
# syntax : while condition:
```

```python
number = int (input("enter the number") )
i=1
while i<11: # work until this condition is true
    print(number*i)
    i+=1       
```

### Continue
- Unlike break , continue statement forces the next iteration of the loop to take place, skipping any code in between.
With the continue statement we can stop the current iteration, and continue with the next:
- It skip the current iteration


```python
 for i in range(10):
     if i == 6:
         continue
     else:
         print(i)
```
