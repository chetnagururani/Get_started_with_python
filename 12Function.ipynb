# Filter
- filter(function, iterable) filters out values from a given iterable that satisfy a condition given inside a function.
  Here we filter out the even numbers from a list of numbers 

```python
def is_even(num):
    return num % 2 == 0

numbers = [1, 2, 3, 4, 5]
evens = filter(is_even, numbers)
print(list(evens))
```

# Recursion(function called inside function(self-invocation)
- which means a defined function can call itself.This has the benefit of meaning that you can loop through data to reach a result.
- The developer should be very careful with recursion as it can be quite easy to slip into writing a function which never terminates

```python
def tri_recursion(k):
  if(k > 0): # this is the stopping condition like in while
    result = k + tri_recursion(k - 1)
    print(result)
  else:
    result = 0
  return result

print("\n\nRecursion Example Results")
tri_recursion(6)
```
