# Write OOP's classes to handle the following scenarios
- A user can create and view 2D coordinates
- A user can find out the distance between 2 coordinates
- A user can find the distance of a coordinate from the origin 
- A user can check if a point lies on a given line
- A user can find the distance between a given 2D point line`

```python
class Point:

    def __init__(self, x, y):# parameterized constructor(we give value through object)
        self.x_cod = x
        self.y_cod = y
        
    def __str__(self):
        return '<{},{}>'.format(self.x_cod, self.y_cod )

    # user can find out the distance between 2 coordinates
    # √[(x2 – x1)^2 + (y2 – y1)^2].
    def euclidean_distance(self,other): # self= (x1, x2), other=(x2,  y2)
        return ( (self.x_cod - other.x_cod)**2+ (self.y_cod - other.y_cod)**2)**0.5

    # user can find the distance of a coordinate from the origin
    def distance_from_origin(self):
        return (self.x_cod**2 + self.y_cod**2)**0.5
        return self.euclidean_distance(Point(0,0)) # Point(0,0) We created object here a shortcut method and self parameter is not need to be give

class Line:
    def __init__(self, A, B, C):
        self.A= A
        self.B= B
        self.C = C

    def __str__(self):
        return '{}x + {}y + {} =0'.format(self.A, self.B, self.C)


    def point_on_line(line,point):
        if line.A*point.x_cod + line.B*point.y_cod + line.C == 0:
              return "lies on the line"
        else:
              return "does not lie on the line"

    def shortest_distance(line,point):
        return abs(line.A*point.x_cod + line.B*point.y_cod + line.C)/(line.A**2 + line.B**2)**0.5


    
    def intersect(self, other):
        # Check if lines are parallel
        if (self.A * other.B == self.B * other.A): #m1=m2 slop are equal checking
            print("The lines are parallel.")
        else:
            # Calculate the intersection point
            x = (self.B * other.C - other.B * self.C) / (self.A * other.B - other.A * self.B)
            y = (self.C * other.A - other.C * self.A) / (self.A * other.B - other.A * self.B)
            print(f"The lines intersect at ({x}, {y}).")
    
    
    # def intersect(self, other):
    #     if (self.A/self.B) != (other.A/other.B):
            
    #         x= (self.B * other.C - other.B * self.C)/(self.A*other.B - other.A* self.B) 
    #         y = (self.C * other.A - other.C * self.A) / (self.A * other.B - other.A * self.B)
    #         a = x / (self.B * other.C - other.B * self.C)
    #         b = -y / (self.A * other.C - other.A * self.C)
    #         c = 1 / (self.A * other.B - other.A * self.B)
    #         if a == b == c:
    #             print("line is intersacting")
    #         else:
    #             print("line is not interacting")
    #     else:
    #         print("they are parallel")
    

```
