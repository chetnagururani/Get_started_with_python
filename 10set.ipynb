# sets
- A set is a collection which is unordered, unchangeable*, and unindexed.
1. unordered- it does not maintain the order in which items are inserted.
2. unindexed - set items are unindexed
3. unique - there cannot be two are more similar item in the set
4. Set item must be immutable(string, int, float, boolean, complex)
5. Set is mutable

```python
a={1, 2, "hello", 4.5}
b={1, 2, "hello", 4.5, True} # True=1
print(b)
```

```python
# The set() Constructor
thisset = set(("apple", "banana", "cherry")) # note the double round-brackets
print(thisset)
```

```python
# Set items are unchangeable, but you can remove items and add new items.
```

<b>2. .intersection()
- The intersection() method will return a new set, that only contains the items present in both sets.
- You can use the & operator instead of the intersection() method

```python
a = {"Mumbai", "Delhi", "Bengaluru", "Chennai", "Kolkata",1, True , 2}
b = {1, True, False,"Hyderabad", "Ahmedabad", "Pune", "Jaipur", "Surat","Bengaluru", "Chennai", "Kolkata"}
cities=a.intersection(b)
print(a.intersection(b))
print(cities)
print(a)
```

<b>3- .intersection_update()
- .intersection_update() method will also keep ONLY the duplicates, but it will change the original set instead of returning a new set.

```python
a = {"Mumbai", "Delhi", "Bengaluru", "Chennai", "Kolkata",1, True , 2}
b = {1, True, False,"Hyderabad", "Ahmedabad", "Pune", "Jaipur", "Surat","Bengaluru", "Chennai", "Kolkata"}
cities=a.intersection_update(b)
print(a)
print(cities)
```

```python
#for updating the intersection
a={1, 3, 4, 5, 6}
b={3, 5, 2, 0, 89}
a.intersection_update(b)
print(a)
```

```python
# The values True and 1 are considered the same value. The same goes for False and 0.
```

<b>4- .difference()
- The difference() method will return a new set that will contain only the items from the first set that are not present in the other set.

```python
a = {"Mumbai", "Delhi", "Bengaluru", "Chennai", "Kolkata",1, True , 2}
b = {1, True, False,"Hyderabad", "Ahmedabad", "Pune", "Jaipur", "Surat","Bengaluru", "Chennai", "Kolkata"}
cities=a.difference(b)
print(cities)
print(a)
```

<b>5- .difference_update
- The difference_update() method will also keep the items from the first set that are not in the other set, but it will change the original set instead of returning a new set.

```python
a = {"Mumbai", "Delhi", "Bengaluru", "Chennai", "Kolkata",1, True, 2}
b = {1, True, False, "Hyderabad", "Ahmedabad", "Pune", "Jaipur", "Surat"}
a.difference_update(b)
a
```

```python
a={1, 34, "hello", True, 1, True}
b={1, 2, 4, 34}
a.difference_update(b)
a
```

<b>6- .symmetric_difference()
- The symmetric_difference() method will keep only the elements that are NOT present in both sets.
- You can use the ^ operator instead of the symmetric_difference() method, and you will get the same result.

```python
a={"hello", True, False, 3, 34}
b={1, 2, 3}
a.symmetric_difference(b)
```

<b>7- .symmetric_difference_update()
- The symmetric_difference_update() method will also keep all but the duplicates, but it will change the original set instead of returning a new set.

```python
a={"hello", True, False, 3, 34}
b={1, 2, 3}
a.symmetric_difference_update(b)
a
```
