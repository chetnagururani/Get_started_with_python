# File handling
File handling in Python allows you to read, write, and manipulate files. You can work with different types of files such as text files (.txt), CSV files (.csv), JSON files (.json), and more.

### Basic File Handling Operations
Python provides the built-in open() function to handle files.

Modes:

"r" – Read (default mode)

"w" – Write (overwrites existing content)

"a" – Append (adds content without overwriting)

"x" – Create (fails if file already exists)

"b" – Binary mode (e.g., "rb" for reading binary files)

- Type of data used in I/O:
    - Test -'1234' as a sequence of unicode chars
    - Binary -12345 as sequence of bytes of its binary equivalent
- Hence there are 2 file type to deal with 
    - Text files -All programming files are text file
    - Binary files -Images , music, video, exe files

### How File I/O is done in most programming language
- Open file
- Write/ Read data
- Close the file

```python
f = open('file.txt', 'w') # Python load this code in RAM in buffer memory
f.write("Hello! how are you")
f.close() # Python remove this code from RAM
f.write("hello") # I/O operation on closed file.
```

# Using Context Manager(With)
- It's good idea to close a file after usage as it will free up the resources
- If we dont close it , grabage collector would close it
- with keyword closes the file as soon as the usage is over

```python
# with
with open('sample1.txt', 'w') as f :
    f.write('hii machine')
```

# serialization and Deserialization
- serialization - process of converting python data type to JSON format
- Deserialization - process of converting JSON to python data type
  

### What is JSON?(JavaScript Object Notation)
- It is a text-based format for storing and exchanging data. It's commonly used for web applications, data analysis, and software engineering


```python
# serialization using json module
import json

l = [1, 2, 3, 4]
with open('demo.json', 'w') as f:
    json.dump(l, f)

```

```python
import json
d = {
    'name' : 'chetna', 
    'age' : 33, 
    'gender' : 'female'}
with open('demo.json', 'w') as f:
    json.dump(d, f, indent=4)
```

```python
# deserialization
import json
with open('demo.json' , 'r') as f:
    d = json.load(f)
    print(d)
    print(type(d))
```

### serializing and deserializing custom object


```python
class Person:
    def __init__(self, fname, lname, age, gender):
        self.fname = fname
        self.lname = lname
        self.age = age
        self.gender = gender
     
```

```python
person = Person('chetna', 'gurur', 20, 'male')
```

```python
# import json
# with open('demo.json', 'w') as f:
#     json.dump(person, f)
# TypeError: Object of type Person is not JSON serializable

import json
def show_object(person):
    if isinstance(person, Person):
        return "{}{} age -> {} gender -> {}".format(person.fname,person.lname,  person.age, person.gender)
with open('demo.json', 'w') as f:
    json.dump(person, f, default= show_object)
```

## Pickling
- Pickling is the process whereby a Python object hierarchy is converted into a byte stream, and unpickling is the inverse operation, whereby a byte stream (from a binary file or bytes-like object) is converted back into an object hierarchy.
- by doing this object as it is comes to the file and can perform all the function as it is
- Pickling in Python refers to the process of serializing and saving an object to a file, so it can be loaded and used later. Serialization converts the object into a byte stream, allowing it to be stored or transmitted.

##### How Pickling Works:
- Python uses the pickle module for pickling and unpickling.
- When pickled, an object is saved in a binary format.
- You can "unpickle" the object to load it back into memory.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def display_info(self):
        print('Hi my name is ', self.name, 'and I am', self.age, 'years old')
```

```python
p = Person('chin', 23)
```

```python
# converting into binary
import pickle
with open('person.pkl', 'wb') as f: # write binary
    p = pickle.dump(p, f)
```

```python
# pickle load
import pickle
with open('person.pkl','rb') as f:
  p = pickle.load(f)

p.display_info()
```

# Pickle Vs Json
Pickle lets the user to store data in binary format. JSON lets the user store data in a human-readable text format.
