# Exception Handling in Python
Exception handling in Python is a mechanism that allows you to handle errors gracefully without crashing your program. Instead of your program terminating on an error, you can catch the exception, handle it, and continue execution.

There are 2 stages where error may happen in program
- During compilation(When converting high level code into Machine understandable code) -> <b>Syntax Error
- During execution(Error when running in RAM) -> <b> Exceptions

### Syntax Error
- Something in the program is not written according to the program's grammar.
- Error is raised by the interpreter/compiler
- You can solve it by rectifying the program

#### Example of syntax error
- Leaving symbols like colon,brackets
- Misspelling a keyword
- Incorrect indentation
- empty if/else/loop/class/function

```python
# catching specific exception
try:
    f = open('sample.txt', 'r')
    print(f.read())
    print(m)
    print(5/0)
    
except FileNotFoundError:
    print('file not found ')
except NameError:
    print('variable not define')
except ZeroDivisonError:
    print("Can't divide by 0")
except Exception as e:
    print(e.with_traceback)# it should be in last
    
```

```python
# else
try:
    f = open('sample.txt', 'r')
except FileNotFoundError:
    print('Not found')
except Exception as e:
    print(e)
else: # when try code is succesfully excecute
    print(f.read())
```

```python
# finally
try:
    f = open('sample.txt', 'r')
except FileNotFoundError:
    print('Not found')
except Exception as e:
    print(e)
else: # when try code is succesfully excecute
    print(f.read())
finally:
    print("this have to print")
```

```python
class Bank:
    def __init__(self, balance):
        self.balance = balance
    def withdraw(self, amount):
        if amount < 0 :
            raise Exception('amount cannot be -ve')
        if self.balance < amount:
            raise Exception('Bank balance is less')
            
        self.balance = self.balance - amount
obj = Bank(10000)
try:
    obj.withdraw(15000)
except Exception as e:
    print(e)
else:
    print(obj.balance)
```

```python
# Creating own custom error class
class MyException(Exception): #You Have to inherit Exception class
    def __init__(self, message):
        print('message')
        
class Bank:
    def __init__(self, balance):
        self.balance = balance
    def withdraw(self, amount):
        if amount< 0:
            raise MyException('amount cannot be -ve')
        if self.balance < amount:
            raise MyException('Bank balance is less')
            
        self.balance = self.balance - amount
obj = Bank(10000)
try:
    obj.withdraw(15000)
except Exception as e:
    print(e)
else:
    print(obj.balance)


```
