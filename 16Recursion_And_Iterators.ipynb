# Recursion
- Recursion is a programming technique where a function calls itself in order to solve a problem. It breaks a complex problem into smaller subproblems of the same type until a base case is reached.

- Key Components of Recursion

1. Base Case: The condition that stops the recursion (prevents infinite loops).
2. Recursive Case: The part of the function where it calls itself with a smaller problem.

```python
# loop
def multiply(a, b):
    result = 0
    for i in range(0, b):
        result += a 
    print(a*b)
multiply(4, 5)
```

```python
# Recursion
def mul(a,b):
    if b == 1:
        return a
    return a + mul(a,(b-1))
mul(3, 4)
```

In recursion, "Last In, First Out" (LIFO) refers to how function calls are stored and retrieved from the call stack. This behavior is similar to a stack data structure, where the last function call made is the first one to be completed (executed and popped off the stack).

```python
def fact(v):
    if v == 1: # base case
        return 1
    return v*fact(v-1)
fact(5)
```

```python
def pal(c):
    if len(c) <= 1:
        print("It is the palindrom")
    else:
        if c[0] == c[-1]:
            pal(c[1: -1])
        else:
            print("Not a palindrome")
pal("malayalam")
pal("python")
```

```python
import time

def fabi(x):
    if x == 0 or x == 1:
        return 1
    return fabi(x-1) + fabi(x-2)
start = time.time()
fabi(36)
print(time.time()-start)
# is complexity is 2^n
```

```python
# as we have calculated the fib(3) and we are again calculated fib(3) for reusing it 
# we can do dynamic coding
import time 

start = time.time()
def memo(m, d):
    if m in d:
        return d[m]
    else:
        d[m]=memo(m-1,d) + memo(m-2,d)
        return d[m]
d = {0:1, 1:1}
print(memo(48, d))
print(time.time()- start)
```

# Power set


```python
def sets(s, index=0):
    if index == len(s):  
        return [[]]
    
    subsets = sets(s, index+1)  # Recursively get subsets of the remaining elements
    num = s[index]
    
    # Create new subsets by adding the current element to each existing subset
    new_subsets = [subset + [num] for subset in subsets]
    
    return subsets + new_subsets  # Combine subsets without and with the current element

# Example usage
s = [1, 2, 3]
result = sets(s)
print(result)
```

# Iterator
- An iterator in Python is an object that allows you to iterate over a collection (such as lists, tuples, dictionaries, and sets) one element at a time.
  
- Iterable: Iterable is an object, that one can iterate over, It generates an iterator when passed to the iter() method. An object that can return an iterator (e.g., lists, tuples, sets).
  
- Iterator: BY which iteration is possible,We use a loop internally there is an iterator object that gives an item to the loop fetching items from a sequence, An object with methods __iter__() and __next__() that returns elements one by one.
    - The advantage of an Iterator is, that it does not take the whole sequence in memory instead it takes one data at a time and performs an operation, and then removes it.
  
- Iteration: Process of traversing one by one in the object, It generates an iterator when passed to the iter() method.It is the general term for taking each item of something, on one after another.The process of retrieving items one by one using an iterator

```python
def Creating_for_loop(iterable):
    iterator = iter(iterable)
    while True:
        try:
            print(next(iterator))
            
        except StopIteration:
            break
        
```

```python
a = [1, 2, 3]
b = range(1, 11)
c = (1, 2,3)
d = {1, 2, 4}
e = {0:1, 3:4, 5:2}
Creating_for_loop(e)
```
