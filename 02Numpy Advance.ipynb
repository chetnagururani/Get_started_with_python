<b> Broadcasting

Broadcasting in NumPy is a powerful mechanism that allows arrays of different shapes to be combined for mathematical operations. 


- The term broadcasting describes hor Numpy treats an array with shape during an arithmetic operation.
- The smaller array is "broadcast across the large array so that they have compatible shapes.


```python
# same shape
a= np.arange(6).reshape(2, 3)
b=np.arange(6, 12).reshape(2, 3)
print(a)
print(b)
print(a+b) # we can add it easily
```

```python
# diff shape
a= np.arange(6).reshape(2, 3)
b=np.arange(3).reshape(1, 3)
print(a)
print(b)

print(a+b) # We can add two different shapes with the help of broadcasting
```

```python
# The smaller array is "broadcast across the large array so that they 
# have compatible shapes.
a= np.arange(6).reshape(2, 3)
b=np.arange(3).reshape(1, 3)
print(a)
print(b)
```

<b> Broadcasting Rule
1. Make the two array have the same dimension.
    - If the number of dimension two array are different , add new dimension with size  1 to the head of the array with the smaller dimension.
2. Making each dimension of the two arrays the same size.
   
    - If the size of each dimension of the two arrays do not match, dimension with size 1 are stretched to the size od the other array.
    - If there is a dimension whose size is not 1 in either of the two array, it cannot be broadcasted, and an error is raised.

```python
import numpy as np
a= np.arange(12).reshape(3, 4)
b= np.arange(3)
print(a)
print(b)
```

```python
a= np.arange(3).reshape(1, 3)
b= np.arange(3).reshape(3, 1)
print(a)
print(b)
a+b
```

```python
import numpy as np
a= np.arange(12).reshape(3, 4)
b= np.arange(12).reshape(4, 3)
a+b
```

broadcast is used in vectorization(Instead of using loop it use broadcasting)

broadcasting plays a crucial role in vectorization, a technique used to optimize computations by eliminating explicit loops and performing operations on entire arrays simultaneously.

```python
# Loss function : tell which thing need to control to reduce error
# In linear Regression : Mean square error(loss function)

# taking example
actual = np.random.randint(1, 50, 25)
predicted = np.random.randint(1, 50, 25)
```

```python
def mse(actual, predicted):
    return np.mean((actual-predicted)**2)
mse(actual, predicted)
```

```python
# Binary cross Entropy(Logistic regression :loss function)
a=np.array([1, 0, 0, 1, 1, 0, 0, 1,0])
count=0
for i in a:
    if i == 1:
        count+=1
p = count/ len(a)
p0 = 1- p
def BCE(a,p, p0):
    return - np.mean(((a)*np.log(p))+((1-(a))* np.log(p0)))
BCE(a,p, p0)
```
