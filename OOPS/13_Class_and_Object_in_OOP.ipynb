# Key Concepts of OOP:
<b>1. Class:
- A class is a blueprint for creating objects. It defines a set of attributes and methods that the objects created from the class will have.

```python
class Car: # keyword and class name
    # class name can written in Pascal case(HelloWorld)
    # Now we create a constructor to create a variable/attribute
    def __init__(self, brand, model):# properties
        self.brand = brand# attribute(Attributes are details that describe objects.)
        self.model = model# attribute

    def display_info(self): # method
        print(f"This car is a {self.brand} {self.model}")
```

<b>2- Object
- An object is an instance of a class. It is a concrete entity based on a class definition, and it can use the attributes and methods defined in the class.

# Creating own data type

### Magic method/ dunder function
- They allow you to define or customize the behavior of objects for specific operations.
- These are predefined methods in Python that start and end with double underscores (e.g., __init__, __str__). 

```python
# Fraction Datatype
```

```python
class Fraction:
    # 1 magic method/ dunder function
    # parameterized constructor : it need input
    def __init__(self, x, y): # Called when an object is created. Used to initialize object attributes.
        self.num = x
        self.den = y

    # 2 nd dunder function
    # Defines the string representation of an object (used by print()).
    def  __str__(self):
        # print("idhar") it not print
        # return 'idhar'
        return '{}/{}'.format(self.num, self.den)

    # 3rd magic method
    def __add__ (self, other):  # there are two parameter self and other 1st fraction object(self) and 2nd (other) 
        # print('idhar')
        new_num = self.num * other.den + other.num * self.den
        new_den = self.den * other.den
        return '{}/{}'.format(new_num, new_den)
        
    # 4th magic method
    def __sub__ (self, other):  # there are two parameter self and other 1st fraction object(self) and 2nd (other) 
        # print('idhar')
        new_num = self.num * other.den - other.num * self.den
        new_den = self.den * other.den
        return '{}/{}'.format(new_num, new_den)
    
    def __mul__ (self, other):  # there are two parameter self and other 1st fraction object(self) and 2nd (other) 
        # print('idhar')
        new_num = self.num * other.num
        new_den = self.den * other.den
        return '{}/{}'.format(new_num, new_den)
        
    def __truediv__ (self, other):  # there are two parameter self and other 1st fraction object(self) and 2nd (other) 
        # print('idhar')
        if other.num == 0:
            raise ZeroDivisionError("Cannot divide by a fraction with a numerator of zero")
        new_num = self.num * other.den
        new_den = self.den * other.num
        return '{}/{}'.format(new_num, new_den)

    # make none magic method
    def convert_to_decimal(self):
        return self.num/self.den
                           
```

```python
fr1= Fraction(3, 4)
print(fr1)
fr2= Fraction(8, 5)
print(fr2)
```
