# Tuple
- Comma separated value are called tuple
- A tuple is a collection which is ordered and unchangeable.
- Tuples are written with round brackets.
- Tuple items are ordered, unchangeable, and allow duplicate values.
- can contain different data type.

Tuple items are indexed, the first item has index [0], the second item has index [1] etc.

```python
a=(1, 2, 2)
```

1- Tuples are unchangeable, meaning that you cannot change, add, or remove items once the tuple is created.

2-  there are some workarounds.

## Change Tuple Values
 Once a tuple is created, you cannot change its values. Tuples are unchangeable, or immutable as it also is called.

But there is a workaround. You can convert the tuple into a list, change the list, and convert the list back into a tuple.

```python
a=("apple", "kiwi" ,"grape")
a[1]='banana'
```

```python
a=("apple", "kiwi", "grape")
y=list(a)
y[1]="banana"
y
```

# Unpacking a Tuple
When we create a tuple, we normally assign values to it. This is called "packing" a tuple:

```python
fruits = ("apple", "banana", "cherry")
fruits=(green, yellow, red) # assigning variable

print(green)
print(yellow)
print(red)
```

```python
fruits = ("apple", "banana", "cherry","banana" , "pinapple")
fruits=(green, yellow, red) # assigning variable

print(green)
print(yellow)
print(red)
```

<b> Join and multiply

```python
fruits = ("apple", "banana", "cherry")
fr1=("banana" , "pinapple")
fruits+ fr1
fruits*3
```

```python
fruits = ["apple", "banana", "cherry"] # in list
fruits*3
```

```python
fruits = ("apple", "banana", "cherry")
fr1=("banana", "cherry")

print(fr1 in fruits )# membership Operator
print("cherry" in fruits)
print(("cherry", "banana") in fruits)
print("cherry","banana"in fruits)
```

```python
fruits = ["apple", "banana", "cherry"]
fr1=["banana", "cherry"]
print(fr1 in fruits )
```

```python
# identity operator
fruits = ("apple", "banana", "cherry")
fruit = ("apple", "banana", "cherry")
print(id(fruit), id(fruits))
print(fruits is fruit)
# for list
fruits1 = ["apple", "banana", "cherry"]
fruit1 = ["apple", "banana", "cherry"]
print(id(fruit1), id(fruits1))
print(fruits1 is fruit1)
```
