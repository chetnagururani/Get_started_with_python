<b> Aliasing
- Aliasing in Python occurs when two or more variables refer to the same object in memory. 
- This means changes made to the object using one alias will be reflected in all the other aliases as they point 
  to the same memory location.
- In Python, variables are references (or pointers) to objects.
When you assign one variable to another without creating a new object, both variables refer to the same object.

# Pass by Reference (pass-by-object-reference)

<b>Pass by object reference:

- Means if we are passing an object in function , you are passing object reference/address.
- Changes to the existing object’s attributes would modify the original.
- Creating a new object (as in this case) breaks the link with the original, leaving it unaffected.
  
<b>Pass by Reference:

- Python mimics this for mutable objects when modifying them directly.
- The original object is affected because both the function and caller reference the same memory( because the function gets access to the same object).

<b>Pass by Value:

- Python mimics this for immutable objects or when reassigning the variable inside the function.
- Any changes create a new object, leaving the original untouched.

#### <b> getter and setter
- getters(get variable) and setters(to modify the variable) are methods used to access and update the attributes of a class, respectively.
-  They provide controlled access to the class attributes, ensuring data encapsulation and validation.
- <b> Getter:</b> Used to retrieve the value of an attribute.
- <b>Setter: <B> Used to set or modify the value of an attribute.

<b> Why Use Getters and Setters?
- To add logic when accessing or modifying attributes (e.g., validation, logging).
- To control how attributes are accessed or updated without directly exposing them.
- To maintain backward compatibility if the internal implementation changes.

### Using Getter Methods
- You can create a public method (like a getter) in the class that calls the private method. This way, you are still respecting the encapsulation.

```python
class MyClass:
    def __init__(self, value):
        self.__value = value 
    def __private_method(self): 
        return f"Value is {self.__value}" 
    def get_value(self): 
        return self.__private_method() # Public method to access private method 
# Create an instance of MyClass 
        
obj = MyClass(10)
print(obj.get_value())
```

### Using Python's @property Decorator
- Python provides a cleaner and more Pythonic way to implement getters and setters using the @property decorator.
- This avoids explicitly calling get_ or set_ methods, making the code more readable.

```python
class Person:
    def __init__(self, name, age):
        self.__name = name  # Private variable
        self.__age = age    # Private variable

    # Getter for name
    @property
    def name(self):
        return self.__name

    # Setter for name
    @name.setter
    def name(self, name):
        if isinstance(name, str):
            self.__name = name
        else:
            print("Name must be a string!")

    # Getter for age
    @property
    def age(self):
        return self.__age

    # Setter for age
    @age.setter
    def age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Age must be positive!")

    @property
    def greet(self):
        print("how are you ",self.name)

# Example usage
person = Person("Alice", 25)

# Accessing private variables
print(person.name)  # Output: Alice
print(person.age)   # Output: 25

# Modifying private variables
person.name = "Bob"
print(person.name)  # Output: Bob

person.age = 30
print(person.age)   # Output: 30

# Invalid input
person.age = -5  # Output: Age must be positive!

person.greet # we can write of function not even need to call it
```

- @property decorator in Python, it allows you to access an attribute as if it were a public variable
- But behind the scenes, it's calling a method.
- This provides a way to run custom logic when getting or setting the attribute, without making it explicit that you are calling a method.
