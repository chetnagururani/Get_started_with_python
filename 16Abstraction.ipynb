# Abstraction(Hidden)
- abstraction means showing only the important details and hiding the complex parts. It helps to focus on what something does rather than how it does it.
- For example on a laptop, there are various operations are performed but they are hidden
- abstraction lets you use something without understanding its complicated inner workings.
- Bank app which has a database class and there are two child class web apps and  mobile apps but a senior developer put a constrain that there should be  a  security code then only they can be child class.

1. Abstract Class:

-  It is meant to be subclassed, and it typically includes abstract methods that must be implemented by subclasses.
-   Abstract classes are useful when you want to define a common interface for a group of related classes.

2. Abstract Method:

- An abstract method is a method that is declared in an abstract class there is no code written
-  Subclasses of the abstract class are required to provide an implementation for these methods.

```python
# abstract class: ,the class which have one abstract method( where the code is not written
from abc import ABC, abstractmethod # ABC abstract base class
class BankApp(ABC): # it inherit from ABC class
    def database(self):
        print("connected to database")

    @abstractmethod # decorator
    def security(self):
        pass

    @abstractmethod # decorator
    def display(self):
        pass
  
```

```python
class MobileApp(BankApp): # to connect the database if we want to inherit
    # For the database class we have to add a security method in our class neither it will gives error
    def mobile_login(self):
        print('login into mobile')
    
    # Adding security method
    def security(self):
        print('mobile security')

    def display(self):
        print("display")
```

```python
mob= MobileApp()
```
